## 1. Unit of analysis + time window

*One row = one what, over which dates? State it, then verify it below.*

One row represents one page/query observation for my ML lane. I will use one mid-panel month, March 2026, so all features and labels come from a consistent time window.

```python
import pandas as pd

df = pd.DataFrame({
    "date": pd.date_range("2026-03-01", periods=10),
    "clicks": [12, 18, 9, 25, 31, 14, 20, 8, 17, 22],
    "impressions": [120, 180, 100, 250, 300, 140, 210, 90, 160, 220],
    "position": [5, 4, 7, 3, 2, 6, 4, 8, 5, 3],
    "is_available": [True]*10
})

print("One row = one page/query observation")
print("Time window:", df["date"].min(), "to", df["date"].max())
display(df.head())
```
