## 1. Ranked actions + reason codes

*The queue: what to do first, and why, in words a human trusts.*

I rank opportunities by priority score. High-impression items with weak click performance receive the highest priority because they already have visibility but may have room for improvement. Each recommendation includes a reason code so a human can understand why it was selected.

```python
import pandas as pd

queue = pd.DataFrame({
    "item": ["Page A", "Page B", "Page C", "Page D", "Page E"],
    "score": [0.92, 0.84, 0.76, 0.61, 0.48],
    "reason_code": [
        "HIGH_IMPRESSIONS_LOW_CTR",
        "HIGH_IMPRESSIONS_LOW_CTR",
        "POSITION_OPPORTUNITY",
        "POSITION_OPPORTUNITY",
        "MONITOR"
    ],
    "action": [
        "Optimize title/content",
        "Improve search snippet",
        "Review content",
        "Review content",
        "Monitor"
    ]
})

queue = queue.sort_values("score", ascending=False)
display(queue)
```
