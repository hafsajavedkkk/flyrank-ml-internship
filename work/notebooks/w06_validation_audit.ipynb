## 3. Leakage audit

*The same hunt from Week 3, on your final feature set.*

I checked the final feature set for information that would only be known after the prediction decision or that is directly derived from the target. These fields must be excluded because they could make validation performance artificially high.

```python
features = ["clicks", "impressions", "position"]

leakage_fields = [
    "future_clicks",
    "future_ctr",
    "final_outcome",
    "target",
    "product_flag"
]

found = [x for x in leakage_fields if x in features]

print("Final features:", features)
print("Potential leakage found:", found)

assert len(found) == 0
print("PASS: No known label-derived or future features used.")
```
