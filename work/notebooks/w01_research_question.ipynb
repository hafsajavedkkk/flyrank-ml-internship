## Research Question

**Research Question:**  
Can machine learning predict which students are at risk of poor academic performance based on attendance, study habits, and previous grades?

**Decision:**  
Help teachers and academic advisors identify students who may need additional academic support.

**Action:**  
Teachers can provide tutoring sessions, counseling, personalized study plans, or additional learning resources to students predicted to be at risk.

**Cost of a Wrong Recommendation:**  
If a struggling student is not identified, they may continue to perform poorly or fail courses. If a student is incorrectly identified as at risk, extra academic support may be provided unnecessarily, which uses time and resources.

**Why Machine Learning?**  
Machine learning can analyze multiple factors simultaneously and discover patterns that are difficult to identify manually, helping educators make better-informed decisions.
